```r
library(pheatmap)
library(viridis)
figsize <- function(args_x,args_y){
  options(repr.plot.width=args_x, repr.plot.height=args_y, repr.plot.pointsize=8)
}
hcol <- rev(inferno(90))
overlap = read.csv('cross_species_heatmap_matrix.csv',row.names = 1)
```

```r
library(ggplot2)
```

```r
overlap = overlap[!rownames(overlap) %in% c('M_acc_Microglia'),!colnames(overlap) %in% c('M_acc_Microglia')]
fig <- pheatmap(overlap,color=hcol,breaks=seq(-0.1,1.5,length.out=101),cluster_rows = F,cluster_cols = F,gaps_row=c(5,10,15,20,25),gaps_col=c(5,10,15,20,25),cellwidth = 10,cellheight = 9,border_color = NA)
ggsave(fig,file='heatmap_crossSpecies_noaccMicroglia_no_H_Cor.pdf')
```
