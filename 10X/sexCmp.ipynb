#### ----------------------statistics------------------------

```python
from scipy.stats import chi2_contingency, fisher_exact
from statsmodels.stats.multitest import multipletests

p_values = []
indices = []

t_statistic, p_val = stats.ttest_rel(cellTypePercent_male, cellTypePercent_female)
p_values.append(p_val)
indices.append("Cell proportion")

t_statistic, p_val = stats.ttest_rel( stats.zscore(meanCounts.loc['meanCounts_male']), stats.zscore(meanCounts.loc['meanCounts_female']) )
p_values.append(p_val)
indices.append("Z-scored detect counts")

t_statistic, p_val = stats.ttest_rel( stats.zscore(mean_nGenes.loc['meanGenes_male']),  stats.zscore(mean_nGenes.loc['meanGenes_female']) )
p_values.append(p_val)
indices.append("Z-scored detect genes")


# multicompare corrention（FDR-Benjamini-Hochberg）
# correction method：'fdr_bh' 
reject, p_corrected, _, _ = multipletests(p_values, alpha=0.05, method='fdr_bh')


result = pd.DataFrame({
    "Original p-value": p_values,
    "Adjusted p-value（FDR）": p_corrected,
    "Significance（α=0.05）": [bool(r) for r in reject]
}, index=indices)

print("Paired t-test and FDR adjusted：")
print(result)
```

```python
def plot_expressCorr(plot_data, figTitle, ax=None):
    """
    绘制基因表达相关性散点图
    
    参数:
    plot_data -- DataFrame，包含'Male'和'Female'两行数据，列名为基因名
    figTitle -- 图形标题
    ax -- 可选的Axes对象，如果为None则在当前画布上创建子图
    """
    if ax is None:
        ax = plt.gca()
    
    # 筛选红色和蓝色基因
    blue_gene = []
    red_gene = []
    
    for index in plot_data.columns:
        diff = plot_data.loc['Female', index] - plot_data.loc['Male', index]
        if diff < -1:
            red_gene.append(index)
        elif diff > 1:
            blue_gene.append(index)
    
    # 设置颜色和大小
    color_list = []
    size_list = []
    
    for gene_index in plot_data.columns:
        if gene_index in red_gene:
            color_list.append('red')
            size_list.append(6)
        elif gene_index in blue_gene:
            color_list.append('blue')
            size_list.append(10)
        else:
            color_list.append('grey')
            size_list.append(1)
    
    # 绘制散点图
    ax.scatter(x=plot_data.loc['Male', :], y=plot_data.loc['Female', :], 
               c=color_list, s=size_list)
    
    # 添加文本标注
    texts = [ax.text(plot_data.loc['Male', gene_index], 
                     plot_data.loc['Female', gene_index], 
                     gene_index) for gene_index in red_gene + blue_gene]
    # adjust_text(texts, arrowprops=dict(arrowstyle="-", color='black', lw=0.5))
    
    # 添加参考线
    ax.axline((0, 0), (4, 4), color='black')  # y=x 线
    ax.axline((0, 1), (3, 4), linestyle='--', color='black', linewidth=0.5)  # y=x+1 线
    ax.axline((1, 0), (4, 3), linestyle='--', color='black', linewidth=0.5)  # y=x-1 线
    
    # 设置坐标轴范围
    ax.set_xlim([0, 4])
    ax.set_ylim([0, 4])
    
    # 设置刻度
    ax.set_xticks([0, 1, 2, 3, 4])
    ax.set_yticks([0, 1, 2, 3, 4])
    
    # 设置坐标轴标签
    ax.set_xlabel('Male', size=12)
    ax.set_ylabel('Female', size=12)
    
    # 设置标题
    ax.set_title(figTitle)
    
    return ax
```
